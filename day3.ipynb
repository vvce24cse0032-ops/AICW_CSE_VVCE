```python
from datetime import date
```

```python
class student:
    def __init__(self,name,age):
        self.name=name
        self.age=age
    @classmethod
    def calculate_age(cls,name,birthyear):
        return cls(name,date.today().year - birthyear)
    def show(self):
        print(self.name + "'s age is " + str(self.age))
```

```python
age=student('chy', 19)
age.show()
```

```python
by = student.calculate_age('chy',2006)
by.show()
```

```python
#static method
class emp:
    @staticmethod
    def sample(x):
        print('inside static method',x)
    #call static method
emp.sample(10)
#can be called using object
emp1=emp()
emp1.sample(10)
```

<h1>INHERITANCE,ENCAPSULATION AND POLYMORPHISM</h1>

```python
#parent class
class person:
    def person_info(self,name,age):
        print('inside person class')
        print('name',name,'age',age)
#parent class 2
class company:
    def company_info(self,company_name,location):
        print('inside the company class')
        print('name',company_name,'location',location)
```

```python
#child class
class employee(person,company):
    def employee_info(salary,skills):
        print('inside employee class-child class')
        print('salary',salary,'skills',skills)
```

```python
#create an object
emp2=employee()
```

```python
#access the data
emp2.person_info('chy',28)
emp2.company_info('edunet','bengaluru')
emp2.employee_info(24000,'aiml')
```
