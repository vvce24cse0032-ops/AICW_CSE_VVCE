```python
#_init_constructor - used to initialize
class person:
    def __init__(self,name,age,school):         #double underscore
        self.name=name
        self.age=age            #instance variables
        self.school=school
    def myfun(self):
        print('hello,my name is',self.name)
        print('my age is',self.age)
        print('i went to',self.school)
```

```python
p1=person('chyy',38,'navkis')
p1.myfun()
```

```python
class student:
    def __init__(self,usn,name,age):
    #instance variable
      self.usn=usn
      self.name=name
      self.age=age
    #instance method access instance variable
    def show(self):
       print('usn',self.usn,'name',self.name,'age',self.age)

       #instance method to modify instance variable
    def update(self,usn,age):
       self.usn=usn
       self.age=age
```

```python
#create an object 
stu=student(2023,'chy',16)
stu.show()
```

```python
stu.update(12,38)
```
